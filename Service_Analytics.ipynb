```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))
sns.heatmap(df.select_dtypes(include=['int64','float64']).corr(),
            annot=True,
            cmap='coolwarm')
plt.show()
```

### Correlation Analysis Insights

1. A **strong positive correlation (0.72)** exists between **Value** and **Net Value**, indicating that higher transaction values directly contribute to higher net revenue.

2. Most of the variables exhibit **very weak correlations** (values close to 0), suggesting that they are largely independent of each other.

3. **Tax Amount** shows a weak negative correlation (-0.14) with **Net Value**, indicating that tax variations have minimal influence on the overall net revenue.

4. **Discount Amount** has a very weak negative relationship (-0.09) with **Net Value**, suggesting that discounts do not significantly impact total revenue.

5. **Customer Number**, **Item**, and **Sales Person** show negligible correlations with financial variables, indicating that these factors do not directly influence transaction values.

6. The heatmap reveals that there are **no significant multicollinearity issues** among the variables, which is beneficial for further statistical or predictive modeling.

7. The analysis confirms that **transaction value is the primary driver of net revenue**, while other operational variables have limited influence.

8. Overall, the dataset demonstrates a balanced structure with minimal interdependency among variables.

**Since Value strongly influences Net Value, the organization should focus on increasing high-value transactions and optimizing pricing strategies to maximize revenue.**

```python
import pandas as pd
import matplotlib.pyplot as plt


df['Billing Date'] = pd.to_datetime(df['Billing Date'])


df['Month_Year'] = df['Billing Date'].dt.to_period('M')


monthly_net = df.groupby('Month_Year')['Net Value'].sum()


monthly_net.index = monthly_net.index.astype(str)


plt.figure(figsize=(12, 6))
plt.plot(monthly_net.index, monthly_net.values, marker='o')


for i, value in enumerate(monthly_net.values):
    plt.text(i, value, f'{value:,.0f}', ha='center', va='bottom')

plt.title('Month-wise Total Net Value')
plt.xlabel('Month')
plt.ylabel('Total Net Value')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

# Insights

- Revenue shows seasonal variations across months.
- Peak revenue was generated during August month.
- Lowest revenue was observed during December month.
- Service demand fluctuates throughout the financial year.

```python
import pandas as pd
import matplotlib.pyplot as plt


df['Billing Date'] = pd.to_datetime(df['Billing Date'])


df['Month_Year'] = df['Billing Date'].dt.to_period('M').astype(str)


monthly_material_net = pd.pivot_table(
    df,
    values='Net Value',
    index='Month_Year',
    columns='Material Type',
    aggfunc='sum',
    fill_value=0
)


ax = monthly_material_net.plot(
    kind='bar',
    stacked=True,
    figsize=(14, 7)
)


for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', label_type='center')

plt.title('Month-wise Net Value by Material Type')
plt.xlabel('Month')
plt.ylabel('Net Value')
plt.xticks(rotation=45)
plt.legend(title='Material Type')
plt.tight_layout()
plt.show()
```

### Insights

1. **HAWA (Parts) material type consistently contributed the highest net value across all months**, making it the major revenue-generating material category.

2. **DIEN (Labour)material type recorded the second-highest net value**, indicating a significant contribution to the overall business revenue.

3. **ZEXT (Lubes) material type contributed the least net value** compared to DIEN and HAWA, although it maintained relatively stable performance throughout the year.

4. The overall net value peaked during the month with a total value of approximately **₹3.8 Lakhs**, driven primarily by high sales of HAWA materials.

5. A noticeable decline in total net value was observed during certain months, indicating fluctuations in customer demand and service activities.

6. The revenue distribution shows that **HAWA materials account for nearly half of the total monthly revenue**, highlighting their importance in inventory planning and stock management.

7. Despite monthly fluctuations, all three material types exhibited a consistent demand pattern, suggesting stable business operations throughout the financial year.

8. Since HAWA and DIEN contribute the majority of revenue, maintaining adequate inventory levels for these material categories is essential to avoid stock shortages and ensure uninterrupted service operations.

```python
import matplotlib.pyplot as plt

order_type_count = df['Order Type'].value_counts()

ax = order_type_count.plot(
    kind='bar',
    figsize=(10, 5)
)


for container in ax.containers:
    ax.bar_label(container)

plt.title('Count of Order Types')
plt.xlabel('Order Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Insights

1. **Paid Schedule Service** recorded the highest number of service orders (**9,032**), indicating that most customers prefer paid maintenance services after the warranty period.

2. **LCV Minor Service** accounted for the second-highest number of orders (**7,444**), highlighting the high demand for routine maintenance and minor repair activities.

3. The counts of **Free Service orders** (First, Second, Third, and Fourth Free Services) are significantly lower compared to paid services, suggesting that many vehicles have moved beyond the free service period.

4. Among free service categories, **LCV Third Free Service (337 orders)** has the highest count, followed closely by **Fourth Free Service (315 orders)** and **Second Free Service (310 orders)**.

5. **LCV First Free Service** has the lowest count (**64 orders**), which may indicate a lower number of newly delivered vehicles during the analysis period.

6. **LCV Major Service** recorded only **4 orders**, suggesting that major repair activities are relatively rare compared to regular maintenance services.

7. Overall, the service center workload is predominantly driven by **paid and minor service activities**, emphasizing the importance of these services in generating revenue.

8. The significantly higher volume of paid services indicates strong customer retention and continuous vehicle usage beyond the warranty period.

```python
import pandas as pd
import matplotlib.pyplot as plt


df['Billing Date'] = pd.to_datetime(df['Billing Date'])


df['Month_Year'] = df['Billing Date'].dt.to_period('M').astype(str)


top_5_vehicles = df['Description'].value_counts().head(5).index


filtered_df = df[df['Description'].isin(top_5_vehicles)]


monthly_vehicle = pd.pivot_table(
    filtered_df,
    index='Month_Year',
    columns='Description',
    aggfunc='size',
    fill_value=0
)


ax = monthly_vehicle.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)


for container in ax.containers:
    ax.bar_label(container, label_type='center')

plt.title('Monthly Count of Top 5 Descriptions')
plt.xlabel('Month')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Description')
plt.tight_layout()
plt.show()
```

# Insights

1. **DOST PLUS** consistently recorded the highest service count across most months, indicating that it is the most frequently serviced vehicle model.

2. The service demand for **BADA DOST i4** remained relatively high throughout the year, making it the second most serviced vehicle category.

3. A significant increase in service count was observed for **DOST PLUS** during certain months, reaching a peak count of approximately **572 services**, suggesting higher maintenance requirements or increased vehicle usage during that period.

4. **BADA DOST i3+** consistently recorded the lowest service counts among the top five vehicle models, indicating comparatively lower customer visits.

5. The service frequency of **Dost 1.25** remained stable across all months, showing consistent customer demand.

6. Overall, the chart indicates that **commercial vehicle variants under the DOST family dominate the service center workload**, contributing significantly to total service operations.

7. Monthly fluctuations across all vehicle models suggest the presence of seasonal service demand patterns, which can be utilized for effective workforce and inventory planning.

8. The concentration of service activities among a few vehicle models indicates that spare parts inventory should primarily focus on these high-frequency vehicles to minimize stock shortages.
